```python
import os
from dotenv import load_dotenv
from utils.utils import inspect, print_indented
from scripts.helpers import read_products, EmbeddingClient
import numpy as np
from scipy.spatial import distance

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")
base_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

client = EmbeddingClient(api_key, base_url)

products = read_products()
```

# Finding the most similar product

## Embedding all products

```python
short_descriptions = [product['short_description'] for product in products]

embeddings = client.create_embeddings(short_descriptions)

for i, product in enumerate(products):
    product['embedding'] = embeddings[i]
```

## Embedding a search query

```python
search_text = 'soap'
search_embedding = client.create_embeddings(search_text)[0]
```

```python
distances = []
for product in products:
    dist = distance.cosine(search_embedding, product['embedding'])
    distances.append(dist)
```

```python
min_dist_ind = np.argmin(distances)
print(products[min_dist_ind]['short_description'])
```
