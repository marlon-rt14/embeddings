```python
import os
from dotenv import load_dotenv
from utils.utils import inspect, print_indented
from scripts.helpers import read_products, EmbeddingClient, create_product_text, find_n_closest
from scipy.spatial import distance
import numpy as np

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")
base_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

client = EmbeddingClient(api_key, base_url)

products = read_products()
```

# Recommendation Systems: Based in user history

```python
products = read_products()
query_product = products[-1] # Product to search. The last one in this case

query_product_text = create_product_text(query_product)

# Simulate the history
user_history = products[0:3]
user_history_texts = [create_product_text(product) for product in user_history]

# Filter products to remove any in user_history
products_filtered = [product for product in products if product not in user_history]
products_filtered_texts = [create_product_text(product) for product in products_filtered]
```

```python
query_product_embedding = client.create_embeddings(query_product_text)[0]
product_filtered_embeddings = client.create_embeddings(products_filtered_texts)
```

```python
user_history_embeddings = client.create_embeddings(user_history_texts)
mean_history_embeddings = np.mean(user_history_embeddings, axis=0)
```

```python
hits = find_n_closest(mean_history_embeddings, product_filtered_embeddings, n=3)
```

```python
for hit in hits:
    product = products_filtered[hit['index']]
    print(product['title'])
```
