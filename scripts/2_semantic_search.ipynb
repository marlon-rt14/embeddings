```python
import os
from dotenv import load_dotenv
from utils.utils import inspect, print_indented
from scripts.helpers import read_products, EmbeddingClient, create_product_text, find_n_closest
from scipy.spatial import distance

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")
base_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

client = EmbeddingClient(api_key, base_url)

products = read_products()
```

# Semantic Search

## Enriched Embeddings

```python
products = read_products()
product_texts = [create_product_text(product) for product in products]
```

```python
product_embeddings = client.create_embeddings(product_texts)
```

## Semantic Search

```python
# Create the query vector from query_text
query_text = 'computer'
query_vector = client.create_embeddings(query_text)[0]

# Find the five closest distances
hits = find_n_closest(query_vector, product_embeddings, n=5)

print(f"Search result for {query_text}:")
for hit in hits:
    product = products[hit['index']]
    print(product['title'])
```
